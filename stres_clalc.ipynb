```python
import numpy as np
import pandas as pd
from pathlib import Path

data_dir = Path("Data_export")
specimens = [f"A{i}" for i in range(1, 8)] + [f"S{i}" for i in range(1, 8)]
D = np.zeros(len(specimens))   # [AD1..AD7, SD1..SD7], filled in the loop
data = {}
```

```python
out_dir = Path("data_with_stress")
out_dir.mkdir(exist_ok=True)

for i, name in enumerate(specimens):
    D[i] = float(input(f"Diameter {name[0]}D{name[1:]} for specimen {name} (in): "))
    df = pd.read_csv(data_dir / f"{name}.csv")
    df["Stress (psi)"] = 4 * df["Force (lbf)"] / (np.pi * D[i]**2)
    data[name] = df
    df.to_csv(out_dir / f"{name}.csv", index=False)

print("Diameters:", D)

#save the diameters so material_properties.ipynb (cell 4) can read them
#saved in tensile_test/, not data_with_stress/: material_properties loads every *.csv in that folder as a specimen
diameters = pd.DataFrame({"Specimen": specimens, "d_o (in)": D})   #specimens and D are in the same order, so rows pair by position
diameters.to_csv("diameters.csv", index=False)
```
